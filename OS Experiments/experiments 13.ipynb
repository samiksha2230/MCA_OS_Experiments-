```python
# Disk Scheduling Algorithms: FCFS and SSTF

def fcfs(requests, head):
    total_seek_time = 0
    for request in requests:
        total_seek_time += abs(request - head)
        head = request
    print("FCFS Total Seek Time:", total_seek_time)

def sstf(requests, head):
    total_seek_time = 0
    while requests:
        # Find the request closest to the current head position
        closest = min(requests, key=lambda x: abs(x - head))
        total_seek_time += abs(closest - head)
        head = closest
        requests.remove(closest)
    print("SSTF Total Seek Time:", total_seek_time)

# Sample disk requests and initial head position
disk_requests = [82, 170, 43, 140, 24, 16, 190]
initial_head = 50

# Run both algorithms
fcfs(disk_requests.copy(), initial_head)
sstf(disk_requests.copy(), initial_head)
```
