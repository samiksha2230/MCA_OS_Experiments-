```python
# Banker's Algorithm to Check System Safety
import numpy as np

def is_safe(alloc, maxm, avail):
    # Calculate the need matrix
    need = maxm - alloc
    work = avail.copy()
    finish = [False] * len(alloc)

    while True:
        found = False
        for i in range(len(alloc)):
            if not finish[i] and all(need[i] <= work):
                work += alloc[i]
                finish[i] = True
                found = True
        if not found:
            break
    return all(finish)

# Allocation matrix
alloc = np.array([
    [0, 1, 0],
    [2, 0, 0],
    [3, 0, 2]
])

# Maximum demand matrix
maxm = np.array([
    [7, 5, 3],
    [3, 2, 2],
    [9, 0, 2]
])

# Available resources
avail = np.array([3, 3, 2])

# Check if the system is in a safe state
print("System Safe:", is_safe(alloc, maxm, avail))
```
