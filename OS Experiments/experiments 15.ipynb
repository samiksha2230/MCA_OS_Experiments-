```python
# Simulate Container Scheduling Based on CPU Usage

# Define a list of containers with their CPU requirements
containers = [
    {"name": "C1", "cpu": 2},
    {"name": "C2", "cpu": 1},
    {"name": "C3", "cpu": 3}
]

# Sort containers by CPU usage (ascending)
containers.sort(key=lambda x: x["cpu"])

# Schedule containers based on sorted CPU usage
for container in containers:
    print(f"Scheduling {container['name']} with CPU {container['cpu']}")
```
