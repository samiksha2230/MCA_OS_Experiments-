```python
# Virtual File System Simulation

class VFS:
    def __init__(self):
        self.files = {}

    def create(self, name, content):
        self.files[name] = content

    def read(self, name):
        print(self.files.get(name, "File not found"))

# Create an instance of the virtual file system
vfs = VFS()

# Create a file and read its content
vfs.create("file1.txt", "Hello VFS")
vfs.read("file1.txt")
```
