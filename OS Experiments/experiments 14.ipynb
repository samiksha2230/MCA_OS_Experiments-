```python
# Monitor File Access & Suggest Cache Strategy
from collections import Counter

# Sample file access log
files = ["a.txt", "b.txt", "a.txt", "c.txt", "a.txt", "b.txt"]

# Count frequency of each file access
access_count = Counter(files)

# Display results
print("File Access Frequency:", access_count)
print("Suggested Cache: LRU (frequent repeated access)")
```
