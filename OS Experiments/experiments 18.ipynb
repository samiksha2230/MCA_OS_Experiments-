```python
# Resource Allocation Simulation (Docker-like)

# Define tasks with their CPU requirements
tasks = [
    {"task": "T1", "cpu": 2},
    {"task": "T2", "cpu": 1}
]

# Total available CPU units
total_cpu = 3

# Allocate CPU to tasks if possible
for task in tasks:
    if total_cpu >= task["cpu"]:
        print(task["task"], "allocated")
        total_cpu -= task["cpu"]
    else:
        print(task["task"], "waiting")
```
